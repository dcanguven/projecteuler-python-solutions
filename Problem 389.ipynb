## Initial approach

* start with the mean and variance of the four sided die
* for a fair s sided die the mean is s plus 1 divided by 2
* its variance is s squared minus 1 divided by 12
* each new total is a random sum where the previous result decides how many dice are thrown
* use the law of total variance instead of calculating every possible score
* update the mean with previous mean times the new die mean
* update the variance with both the die variance and the variance of the random number of throws
* repeat this for the 6 sided, 8 sided, 12 sided and 20 sided dice

```python
from fractions import Fraction


def die_stats(sides):
    mean = Fraction(sides + 1, 2)
    variance = Fraction(sides * sides - 1, 12)
    return mean, variance


mean, variance = die_stats(4)

for sides in [6, 8, 12, 20]:
    die_mean, die_variance = die_stats(sides)

    new_variance = mean * die_variance + variance * die_mean ** 2
    new_mean = mean * die_mean

    mean = new_mean
    variance = new_variance
```

```python
%%time
result = round(float(variance), 4)
print("Result:", result)
```
