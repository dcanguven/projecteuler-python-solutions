## Initial approach

- after the first knight sits down, the circular table becomes one long row of available chairs
- study the expected number of occupied chairs for a straight row first
- splitting at the first occupied chair gives a recurrence for smaller rows
- for very large rows this recurrence approaches a fixed limiting density
- the target size is so large that the remaining finite size error cannot affect fourteen decimal places
- compute the limiting fraction of empty chairs and round it to fourteen digits

```python
%%time

import math

N = 10**18

result = (1 + math.exp(-2)) / 2

print("Result:", f"{result:.14f}")
```
