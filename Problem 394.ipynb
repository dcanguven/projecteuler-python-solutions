## Initial approach

-   represent the remaining pie as a fraction of its previous size
-   the later of the two random cuts decides how much pie remains
-   if the cut positions are uniform then the remaining ratio has density two times one minus the ratio
-   take the negative logarithm so repeated multiplication becomes addition
-   the logarithmic decrease has density two times the difference of two exponentials
-   write a renewal equation for the expected number of procedures before passing the limit
-   solve the resulting differential equation to get a direct formula for E of x
-   substitute x equals 40 and round the answer to ten decimal places

```python
import math


def expected_steps(x):
    return (
        7 / 9
        + 2 * math.log(x) / 3
        + 2 / (9 * x ** 3)
    )
```

```python
%%time
assert abs(expected_steps(1) - 1) < 1e-12
assert abs(expected_steps(2) - 1.2676536759) < 1e-10
assert abs(expected_steps(7.5) - 2.1215732071) < 1e-10
result = expected_steps(40)
print("Result:", f"{result:.10f}")
```
