## Initial approach

- there are 70 balls with 10 balls for each of the 7 colours
- calculate the probability that one particular colour does not appear among the 20 selected balls
- this means all 20 balls must come from the other 60 balls
- that probability is C(60, 20) / C(70, 20)
- so the probability that a particular colour appears is one minus that value
- by linearity of expectation multiply this probability by 7
- no simulation is needed because the expectation can be calculated directly with combinations

```python
%%time

from math import comb

total_balls = 70
balls_per_colour = 10
colours = 7
picked = 20

missing_probability = (
    comb(total_balls - balls_per_colour, picked)
    / comb(total_balls, picked)
)

result = colours * (1 - missing_probability)

print("Result:", f"{result:.9f}")
```
