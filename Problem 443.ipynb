## Initial approach

-   start from the useful point `g(9) = 27 = 3 × 9`
-   after a point where `g(a) = 3a` the sequence usually increases only by one for many steps
-   during this linear part the next gcd can be rewritten using the fixed number `2a - 1`
-   find the smallest prime factor `p` of `2a - 1` to locate the next nontrivial jump
-   the next special index is `a + (p - 1) // 2` and at that point the relation `g(a) = 3a` appears again
-   jump directly between these special indices instead of calculating every value up to `10^15`
-   use sympy only to factor `2a - 1` quickly and get its smallest prime factor

```python
%%time

from sympy import factorint

target = 10**15
a = 9

while a < target:
    factors = factorint(2 * a - 1)
    p = min(factors)

    next_a = a + (p - 1) // 2

    if next_a > target:
        result = target + 2 * a
        break

    a = next_a

if a == target:
    result = 3 * a

print("Result:", result)
```
