## Approach

The sequence generated by the pseudo-random number generator eventually becomes periodic.

For each new value, a monotonic increasing stack is maintained. Larger values are removed because they can never become the minimum of any future subsequence ending at the current position.

Each stack entry stores both the value and the number of consecutive starting positions for which that value is currently the minimum. This allows the contribution of the current element to be updated in amortized constant time instead of checking every subsequence individually.

The running contribution is accumulated into the final answer for every processed element.

## Architecture

- generate the pseudo-random sequence until the first repeated state
- maintain a monotonic increasing stack of candidate minimum values
- keep the contribution of all subsequences ending at the current position
- update the answer after processing each new element
- detect the sequence period for future optimization

## Status

**To be optimized**

The current implementation processes only a single period of the generator. The final solution should exploit the periodic structure of the sequence to efficiently handle the required length of `2 × 10^9` elements without iterating over every value.
