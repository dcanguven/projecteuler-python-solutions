## Initial approach

* each tiling step replaces every rectangle with the same smaller pattern
* counting the new four tile meeting points gives a recurrence for the total
* simplifying that recurrence gives a direct expression using powers of two and four
* the required tiling level is far too large to calculate directly
* only the exponent modulo the totient of the final modulus is needed
* compute that reduced exponent with modular exponentiation
* then evaluate the direct expression modulo the required number

```python
%%time

MOD = 17**7
PHI = MOD - MOD // 17

k = 10**18

n_mod = pow(10, k, PHI)

power_2 = pow(2, n_mod, MOD)
power_4 = pow(4, n_mod, MOD)

inverse_15 = pow(15, -1, MOD)

result = (
    (6 * power_4 - 20 * power_2 + 14)
    * inverse_15
) % MOD

print("Result:", result)
```
