## Initial approach

- use the tangency condition to parametrize all integer radius triples
- each primitive solution comes from a coprime pair of positive integers
- the three radii are built from simple squares of these two integers
- every primitive triple can be scaled by any positive integer
- limit the scale so the largest radius stays below one billion
- sum all scaling factors at once with a triangular number
- iterate only over the small range of coprime parameter pairs
- add all valid radius sums

```python
%%time

from math import gcd

LIMIT = 10**9

def triangular(n):
    return n * (n + 1) // 2

result = 0
n = 1

while (n + 1) * (n + 1) * n * n <= LIMIT:
    n2 = n * n

    for m in range(1, n + 1):
        if gcd(m, n) != 1:
            continue

        m2 = m * m
        total2 = (m + n) * (m + n)

        r_b = total2 * n2

        if r_b > LIMIT:
            break

        r_a = total2 * m2
        r_c = m2 * n2

        multiples = LIMIT // r_b

        result += (
            r_a + r_b + r_c
        ) * triangular(multiples)

    n += 1

print("Result:", result)
```
