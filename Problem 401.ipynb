## Initial approach

-   the function g(n) is the sum of the squares of all divisors of n
-   instead of calculating g(n) one by one, reverse the divisor relationship
-   every number d contributes d² to all multiples of d
-   for each divisor d, count how many multiples of d exist up to N
-   the total contribution of d is d² times floor(N / d)
-   sum these contributions for all d from 1 to N
-   use the given modulus because the final answer is extremely large

```python

%%time

N = 10**15
MOD = 10**9

result = 0

for d in range(1, N + 1):
    result = (result + d * d * (N // d)) % MOD

print("Result:", result)
```
