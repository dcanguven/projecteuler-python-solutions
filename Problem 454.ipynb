## Initial approach

-   repeatedly replace the current value with the last nine digits of the power
-   this process quickly reaches a fixed value where applying the operation changes nothing
-   that fixed value is the required trailing digit solution for the current number
-   numbers divisible by ten cannot have a positive solution, so return zero immediately
-   start the iteration from the number itself to reach the fixed point quickly
-   use Python's built in modular power so the huge powers are never created directly
-   calculate this fixed point for every number from two to one million
-   add all fixed points together

```python
%%time

LIMIT = 1_000_000
MOD = 1_000_000_000

def f(n):
    if n % 10 == 0:
        return 0

    x = n

    while True:
        next_x = pow(n, x, MOD)

        if next_x == x:
            return x

        x = next_x

result = sum(f(n) for n in range(2, LIMIT + 1))

print("Result:", result)
```
