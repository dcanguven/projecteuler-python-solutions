## Initial approach

-   track Louise's last number and the amount already accumulated when she finishes
-   the possible stopping positions have a simple density because sums below one correspond to simplex volumes
-   after Louise stops, Julie continues from the current total until the sum passes two
-   for a fixed Louise value, calculate the probability that Julie's final number is larger
-   split the calculation into two regions depending on which value is larger during the remaining process
-   solving these two regions gives a closed form expression for Julie's winning probability
-   use only the math library because the final expression depends on the constant e
-   evaluate the expression and round it to ten decimal places

```python
%%time

import math

e = math.e

result = -5 * e**2 / 4 + 1 / 4 + 7 * e / 2

print("Result:", f"{result:.10f}")
```
